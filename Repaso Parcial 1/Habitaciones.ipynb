```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Datos
habitaciones = [23,47,21,39,37,29,23,44,45,16,30,42,54,
                27,34,15,19,38,44,47,43,38,51,61,39]

ingreso = [1452,1361,1426,1470,1456,1430,1354,1442,1394,1459,
           1399,1458,1537,1425,1445,1439,1348,1450,1431,
           1446,1485,1405,1461,1490,1426]

df = pd.DataFrame({
    "Habitaciones": habitaciones,
    "Ingreso": ingreso
})

df.head()
```

```python
plt.figure()
plt.scatter(df["Habitaciones"], df["Ingreso"])

# Calcular línea de tendencia
x = df["Habitaciones"]
y = df["Ingreso"]

m, b = np.polyfit(x, y, 1)  # pendiente y intercepto
plt.plot(x, m*x + b, linestyle='--')  # línea de tendencia

plt.xlabel("Habitaciones ocupadas")
plt.ylabel("Ingreso restaurante")
plt.title("Gráfica de correlación lineal simple")

plt.show()

# Mostrar ecuación
print(f"Ecuación de la recta: y = {m:.2f}x + {b:.2f}")
```

```python
correlacion = df["Habitaciones"].corr(df["Ingreso"])
correlacion
```

```python
np.corrcoef(df["Habitaciones"], df["Ingreso"])[0,1]
```

# Interpretacion 
Al observar la gráfica de dispersión se nota una tendencia ascendente en los puntos. Esto indica que cuando aumenta el número de habitaciones ocupadas, el ingreso del restaurante también tiende a aumentar.

La relación no es perfecta, ya que los puntos no están completamente alineados, pero sí muestran una correlación positiva moderada.

En conclusión, existe una relación directa entre la ocupación del hotel y el ingreso del restaurante: a mayor número de huéspedes, mayor es el ingreso diario del restaurante.
